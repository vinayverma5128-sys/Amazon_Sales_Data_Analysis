```python
# IMPORTING LIBRARIES
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#LOAD THE DATA
df = pd.read_csv("Amazon.csv")
df.head(25)
```

```python
# Basic Data Exploration
df.shape
```

```python
# Display all column names
df.columns
```

```python
# Check missing values
df.isnull().sum()
```

```python
# Check duplicate rows
df.duplicated().sum()
```

```python
# Get basic statistical information
df.describe()
```

```python
# DATA CLEANING 
# Remove duplicate rows
df = df.drop_duplicates()
```

```python
# Fill missing city names
df["City"] = df["City"].fillna("Unknown")
print(df)
```

```python
# Convert the order date into datetime format
df["OrderDate"] = pd.to_datetime(df["OrderDate"])
print(df)
```

```python
# SALES ANALYSIS USING PANDAS
# Calculate total sales
total_sales = df["TotalAmount"].sum()

print("Total Sales:", total_sales)
```

```python
# Find the top 5 selling products by revenue
top_products = df.groupby("ProductName")["TotalAmount"].sum()

top_products = top_products.sort_values(ascending=False)

print(top_products.head(5))
```

```python
# Find total sales by category
category_sales = df.groupby("Category")["TotalAmount"].sum()

print(category_sales)
```

```python
# Find the most popular payment method
print(df["PaymentMethod"].value_counts().head(1))
```

```python
# Calculate the average order value
average_order = df["TotalAmount"].mean()

print("Average Order Value:", round(average_order, 2))
```

```python
# Find the highest-priced product
highest_price = df.loc[df["UnitPrice"].idxmax()]

print(highest_price[["ProductName", "UnitPrice"]])
```

```python
#Find sales by state
state_sales = df.groupby("State")["TotalAmount"].sum()

print(state_sales.sort_values(ascending=False))
```

```python
# DATA VISUALIIZATION USING MATPLLOTLIB
# Bar chart of category sales
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
category_sales = df.groupby("Category")["TotalAmount"].sum()

category_sales.plot(kind="bar", figsize=(8, 5),color='pink')

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales (INR)")

plt.tight_layout()
plt.show()
```

```python
# Monthly sales analysis
df["Month"] = df["OrderDate"].dt.month

monthly_sales = df.groupby("Month")["TotalAmount"].sum()

monthly_sales.plot(kind="line", marker="o")

plt.title("Monthly Sales Analysis")
plt.xlabel("Month")
plt.ylabel("Total Sales (INR)")

plt.show()
```

```python
# VISULIZATION USING SEABORN
# Product category sales comparison
plt.figure(figsize=(10, 5))

sns.barplot(
    data=df,
    x="Category",
    y="TotalAmount",
    estimator="sum",
    palette="Set2"
)

plt.title("Total Sales by Product Category")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Check the relationship between discount and sales amount
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Discount",
    y="TotalAmount",
    palette="Set2"
)

plt.title("Discount vs Sales Amount")
plt.show()
```

```python
# Create a correlation heatmap
plt.figure(figsize=(8, 5))

sns.heatmap(
    df[[
        "Quantity",
        "UnitPrice",
        "Discount",
        "TotalAmount"
    ]].corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Sales Correlation Heatmap")
plt.show()
```

```python
# Find the top 10 customers by spending
customer_sales = df.groupby("CustomerID")["TotalAmount"].sum()

top_customers = customer_sales.sort_values(ascending=False).head(10)

print(top_customers)
```

```python
# Calculate the percentage of each category's sales
category_sales = df.groupby("Category")["TotalAmount"].sum()

category_percentage = (
    category_sales / category_sales.sum()
) * 100

print(category_percentage.round(2))
```

```python
# Find cancelled orders
cancelled_orders = df[df["OrderStatus"] == "Cancelled"]

print("Total Cancelled Orders:", len(cancelled_orders))

print(cancelled_orders.head())
```

```python
# Final Project Summary
print("========== AMAZON SALES REPORT ==========")

print("Total Orders:", df["OrderID"].nunique())

print("Total Sales:", round(df["TotalAmount"].sum(), 2))

print("Total Products:", df["ProductName"].nunique())

print("Total Customers:", df["CustomerID"].nunique())

print("Total Categories:", df["Category"].nunique())

print("Total Cancelled Orders:",
      (df["OrderStatus"] == "Cancelled").sum())

print("=========================================")
```
